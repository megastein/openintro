#Distributions of random variables

```r
shadenorm = function(below=NULL, above=NULL, pcts = c(0.025,0.975), mu=0, sig=1, numpts = 500, color = "gray", dens = 40,

                    justabove= FALSE, justbelow = FALSE, lines=FALSE,between=NULL,outside=NULL){



    if(is.null(between)){

         below = ifelse(is.null(below), qnorm(pcts[1],mu,sig), below)

         above = ifelse(is.null(above), qnorm(pcts[2],mu,sig), above)

    }



    if(is.null(outside)==FALSE){

         below = min(outside)

         above = max(outside)

    }

    lowlim = mu - 4*sig

    uplim  = mu + 4*sig



    x.grid = seq(lowlim,uplim, length= numpts)

    dens.all = dnorm(x.grid,mean=mu, sd = sig)

    if(lines==FALSE){

          plot(x.grid, dens.all, type="l", xlab="X", ylab="Density")

    }

    if(lines==TRUE){

          lines(x.grid,dens.all)

    }



    if(justabove==FALSE){

        x.below    = x.grid[x.grid<below]

        dens.below = dens.all[x.grid<below]

        polygon(c(x.below,rev(x.below)),c(rep(0,length(x.below)),rev(dens.below)),col=color,density=dens)

    }

    if(justbelow==FALSE){

        x.above    = x.grid[x.grid>above]

        dens.above = dens.all[x.grid>above]

        polygon(c(x.above,rev(x.above)),c(rep(0,length(x.above)),rev(dens.above)),col=color,density=dens)

    }



    if(is.null(between)==FALSE){

         from = min(between)

         to   = max(between)



         x.between    = x.grid[x.grid>from&x.grid<to]

         dens.between = dens.all[x.grid>from&x.grid<to]

         polygon(c(x.between,rev(x.between)),c(rep(0,length(x.between)),rev(dens.between)),col=color,density=dens)

    }



}
```

##4.37 University admissions.

Suppose a university announced that it admitted 2,500 students for the
following year's freshman class. However, the university has dorm room spots for only 1,786 freshman
students. If there is a 70% chance that an admitted student will decide to accept the other and attend this
university, what is the approximate probability that the university will not have enough dormitory room
spots for the freshman class?

```r
n <- 2500
p <- 0.7
available_spots <- 1786

mean <- n * p
sd <- sqrt(n * p * (1 - p))

probability <- 1 - pnorm(available_spots, mean, sd)

probability
```

a. 0.058

##4.42 Survey response rate.

Pew Research reported that the typical response rate to their surveys is only
9%. If for a particular survey 15,000 households are contacted, what is the probability that at least 1,500
will agree to respond?

```r
mu <- n * p
sigma <- sqrt(n * p * (1 - p))

k <- 1500

z <- (k - mu) / sigma

probability <- 1 - pnorm(z)

probability
```

a. 0.00000936

##4.48 Multiple choice quiz.

In a multiple choice quiz there are 5 questions and 4 choices for each question
(a, b, c, d). Robin has not studied for the quiz at all, and decides to randomly guess the answers. What is
the probability that

<ol type = 'a'>
<li>the first question she gets right is the 3rd question?
<li>she gets exactly 3 or exactly 4 questions right?

```r
p_wrong <- 3/4

p_right <- 1/4

p_part1 <- p_wrong * p_wrong * p_right
print(p_part1)
```

```r
p_3right <- dbinom(3, size = 5, prob = p_right)


p_4right <- dbinom(4, size = 5, prob = p_right)

p_part2 <- p_3right + p_4right
print(p_part2)
```
